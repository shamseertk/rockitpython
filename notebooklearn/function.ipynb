```python
# add a method to log one epocj and one to return the best accuracy so far.
class ModelTracker:
  def __init__(self, model_name):
    self.model_name = model_name

  def log_epoch(self, epoch, accuracy):
    if not hasattr(self, 'best_accuracy'):
        self.best_accuracy = accuracy
    else:
        self.best_accuracy = max(self.best_accuracy, accuracy)
    print(f"Epoch: {epoch}, Accuracy: {accuracy}, Best Accuracy so far: {self.best_accuracy}")

  def get_best_accuracy(self):
    return self.best_accuracy

```

```python
#create a tracker for ResNet and log two epocks, then print the best accuracy so far.
class ResNetTracker:
    def __init__(self, model_name="ResNet"):
        self.model_name = model_name

    def log_epoch(self, epoch, accuracy):
        if self.best_accuracy is None:
            self.best_accuracy = accuracy
        else:
            self.best_accuracy = max(self.best_accuracy, accuracy)
        print(f"Epoch: {epoch}, Accuracy: {accuracy}, Best Accuracy so far: {self.best_accuracy}")

    def get_best_accuracy(self):
        return self.best_accuracy

tracker = ResNetTracker()
tracker.log_epoch(1, 0.85)
tracker.log_epoch(2, 0.90)
print("Best Accuracy so far: ", tracker.get_best_accuracy())
```

```python
# A tracker that exists but has not trainet yet. This is not an edge case or a
# hostile input. It is state ever tracker is in for its first few seconds.

fresh = ModelTracker("BERT")
fresh.best_accuracy = getattr(fresh, "best_accuracy", None)
print(f"Best Accuracy for fresh tracker: {fresh.get_best_accuracy()}")
```
